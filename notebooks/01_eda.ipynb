```python
from pathlib import Path
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
sys.path.append(str(ROOT / 'src'))
from preprocess import main as preprocess
from feature_builder import main as build_features

preprocess()
build_features()
df = pd.read_csv(ROOT / 'data' / 'modeling_dataset.csv')
df.head()
```

```python
df[['fan_score', 'a_sponsor_power_index', 'event_attention_m', 'media_reposts_k', 'sponsor_roi']].describe()
```

```python
corr_cols = ['fan_score', 'a_sponsor_power_index', 'team_a_strength', 'event_attention_m', 'media_reposts_k', 'a_player_followers_m', 'sponsor_roi']
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='viridis')
plt.title('Attention, Sponsor Strength and ROI Correlation')
plt.show()
```

研究洞察记录：

- FanScore 越高，说明球迷注意力越强，是 ROI 的关键候选变量。
- Sponsor Power Index 反映品牌是否有能力承接赛事曝光。
- 如果 ROI 只由胜负解释，项目会变成体育预测；如果 ROI 由注意力和 sponsor activation 共同解释，才具备商业研究价值。
