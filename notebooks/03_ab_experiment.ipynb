# 03 Counterfactual A/B Experiment

目标：模拟核心球员缺席、赞助商升级、媒体热度下降三种场景对 ROI 的影响。

```python
from pathlib import Path
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
sys.path.append(str(ROOT / 'src'))
from ab_simulation import main as run_ab

run_ab()
ab = pd.read_csv(ROOT / 'reports' / 'ab_simulation_results.csv')
ab.head()
```

```python
summary = ab.groupby('scenario', as_index=False)['roi_lift_pct'].mean()
sns.barplot(data=summary, x='scenario', y='roi_lift_pct')
plt.xticks(rotation=25, ha='right')
plt.title('Average ROI Lift by Counterfactual Scenario')
plt.show()
summary
```

业务解释：

- 核心球员缺席如果造成显著 ROI 下滑，说明品牌风险集中在明星影响力。
- 赞助商升级如果带来 ROI lift，说明 activation quality 与 brand fit 值得投入。
- 媒体热度下降如果影响最大，说明 sponsor campaign 需要备用传播策略。
